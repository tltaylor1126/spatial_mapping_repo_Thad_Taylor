# 🔬 Visualization Mini-Lab: Trends and Outliers

Now it’s time to put your skills to the test!
In this mini-lab, you’ll explore a real dataset, create visualizations, and identify **trends and outliers**.

We’ll use the famous **Penguins dataset** (thanks to Seaborn).

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
#penguins = sns.load_dataset("penguins")
penguins = pd.read_csv("../../data/penguins.csv")
penguins.head()
```

## 1. Explore Distributions

First, let’s look at the distribution of **body mass**.

```python
sns.histplot(data=penguins, x="body_mass_g", kde=True, hue="species")
plt.title("Distribution of Body Mass by Species")
plt.show()
```

👉 **Task:** Do some species have higher average body mass than others?

## 2. Relationships Between Variables

Scatterplots help us see trends between two quantitative variables.

```python
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="species")
plt.title("Flipper Length vs Body Mass")
plt.show()
```

👉 **Task:** Which species tends to have longer flippers AND higher body mass?

## 3. Box Plots for Outliers

Box plots are great for spotting outliers.

```python
sns.boxplot(data=penguins, x="species", y="bill_length_mm", palette="Set2")
plt.title("Bill Length by Species")
plt.show()
```

👉 **Task:** Can you spot any outliers in the bill length data?

## 4. Correlation Heatmap

Correlation heatmaps let us see overall variable relationships.

```python
corr = penguins.dropna().corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.show()
```

👉 **Task:** Which variables appear to be strongly correlated?

```python
import pandas as pd
import matplotlib.pyplot as plt

try:
    import seaborn as sns
    viz_df = sns.load_dataset('iris')
except Exception:
    viz_df = pd.DataFrame({
        'sepal_length': [5.1, 4.9, 6.2, 5.9, 7.0, 6.4, 5.5, 4.7, 7.2],
        'sepal_width':  [3.5, 3.0, 3.4, 3.0, 3.2, 2.8, 2.4, 3.2, 3.6],
        'petal_length': [1.4, 1.4, 5.4, 5.1, 4.7, 5.6, 3.8, 1.3, 6.1],
        'petal_width':  [0.2, 0.2, 2.3, 1.8, 1.4, 2.2, 1.1, 0.2, 2.5],
        'species': ['setosa', 'setosa', 'virginica', 'virginica', 'versicolor', 'virginica', 'versicolor', 'setosa', 'virginica'],
    })

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for species, group in viz_df.groupby('species'):
    axes[0].scatter(group['sepal_length'], group['petal_length'], label=species)
axes[0].set_title('Sepal length vs petal length')
axes[0].set_xlabel('Sepal length')
axes[0].set_ylabel('Petal length')
axes[0].legend()

viz_df.groupby('species')['petal_width'].mean().plot(kind='bar', ax=axes[1], color='#2a9d8f')
axes[1].set_title('Average petal width by species')
axes[1].set_ylabel('Petal width')

viz_df.boxplot(column='sepal_width', by='species', ax=axes[2])
axes[2].set_title('Sepal width outlier check')
axes[2].set_xlabel('Species')
axes[2].set_ylabel('Sepal width')
fig.suptitle('')
plt.tight_layout()
plt.show()

print('Summary: petal measurements separate the species more clearly than sepal measurements. The boxplot makes it easier to spot unusually wide or narrow sepals within each species.')
```
