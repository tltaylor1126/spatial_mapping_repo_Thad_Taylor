# 📘 03 - File Loops and `with`

Use `with open(...)` to read files and process content.

## ✅ Learning Goals
- Open and read files using `with`
- Loop over lines in a file
- Use `.strip()` and `.split()` to process text

```
# 📝 Sample file read
with open('sample.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

```
# 🔍 Process each line
with open('sample.txt', 'r') as f:
    for line in f:
        parts = line.strip().split(',')
        print('Parts:', parts)
```

### ✏️ Your Turn
Open a text file and print all lines that contain more than 3 words.

```
from pathlib import Path

path = Path('sample_lines.txt')
path.write_text(
    'short line
'
    'this line has more than three words
'
    'maps are useful
'
    'spatial data analysis is very interesting
',
    encoding='utf-8',
)

with path.open('r', encoding='utf-8') as file:
    for line in file:
        if len(line.split()) > 3:
            print(line.strip())
```
