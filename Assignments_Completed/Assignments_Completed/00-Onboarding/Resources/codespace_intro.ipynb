```python
import json
from pathlib import Path
from IPython.display import JSON, display

path = Path('campus_world.geojson')
if not path.exists():
    sample_geojson = {
        'type': 'FeatureCollection',
        'features': [
            {
                'type': 'Feature',
                'properties': {'name': 'Sample campus point'},
                'geometry': {'type': 'Point', 'coordinates': [-98.4936, 33.8751]},
            }
        ],
    }
    path.write_text(json.dumps(sample_geojson, indent=2), encoding='utf-8')

geojson_data = json.loads(path.read_text(encoding='utf-8'))
display(JSON(geojson_data))
```
