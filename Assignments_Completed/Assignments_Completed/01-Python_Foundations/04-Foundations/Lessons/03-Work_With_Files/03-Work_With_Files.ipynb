✅ **Your Turn**: Navigate into a subfolder of your repo (e.g., `Foundations/`) and list its contents.

```python
from pathlib import Path

current = Path.cwd()
subfolders = [p for p in current.iterdir() if p.is_dir()]
folder = subfolders[0] if subfolders else current
print('Listing:', folder)
for item in sorted(folder.iterdir()):
    print(item.name)
```

## 2. Writing and Reading Text Files

```python
# Writing to a text file
with open("example.txt", "w") as f:
    f.write("Hello, Jupyter!\n")
    f.write("This is a text file.\n")
```

```python
# Reading from a text file
with open("example.txt", "r") as f:
    contents = f.read()
contents
```

## 3. JSON Files

```python
import json

data = {"name": "Graylian", "type": "mascot", "skills": ["howl", "parrot talk"]}

# Write JSON
with open("data.json", "w") as f:
    json.dump(data, f)

# Read JSON
with open("data.json", "r") as f:
    loaded = json.load(f)
loaded
```

✅ **Your Turn**: Create a dictionary with your name and 2-3 favorite courses. Save it to `courses.json`, then read it back.

```python
from pathlib import Path
import json

courses = {
    'name': 'Thad Taylor',
    'favorite_courses': ['Spatial Data and Mapping', 'Data Structures', 'Database Systems'],
}
path = Path('courses.json')
path.write_text(json.dumps(courses, indent=2), encoding='utf-8')
loaded_courses = json.loads(path.read_text(encoding='utf-8'))
print(loaded_courses)
```

## 4. CSV Files

```python
import csv

# Writing CSV
rows = [["Name", "Score"], ["Alice", 90], ["Bob", 85]]
with open("scores.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# Reading CSV
with open("scores.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

✅ **Your Turn**: Create a CSV with 3 rows of your own data (e.g., pets, grades, or hobbies). Read it back and display the rows.

```python
from pathlib import Path
import csv

path = Path('hobbies.csv')
rows = [
    {'hobby': 'mapping', 'hours_per_week': 4},
    {'hobby': 'coding', 'hours_per_week': 6},
    {'hobby': 'reading', 'hours_per_week': 3},
]
with path.open('w', newline='', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=['hobby', 'hours_per_week'])
    writer.writeheader()
    writer.writerows(rows)

with path.open('r', newline='', encoding='utf-8') as file:
    for row in csv.DictReader(file):
        print(row)
```

### Summary
- `%pwd`, `%ls`, `%cd` show you where you are in the filesystem.
- Use `with open(...)` to safely read/write files.
- JSON is for structured data (key/value).
- CSV is for tabular data (rows/columns).
- Managing files is step 1 in any real data science project.
