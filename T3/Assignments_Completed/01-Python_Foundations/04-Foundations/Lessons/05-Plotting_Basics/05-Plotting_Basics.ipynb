# 📈 05 - Plotting Basics

Visualization is a core part of data science.  
In this notebook you will learn:
- Setting up Matplotlib for inline plotting
- Creating simple line, scatter, and bar plots
- Customizing titles, labels, and legends
- Why visuals matter in data storytelling


## 1. Setup

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

```python
import matplotlib.pyplot as plt
x = [0, 1, 2, 3, 4, 5]
y = [n ** 2 for n in x]
plt.plot(x, y, marker="o")
plt.xlabel("x")
plt.ylabel("x squared")
plt.title("Basic Line Plot")
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Line Plot

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.title("Line Plot Example")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.show()
```

✅ **Your Turn**: Plot `cos(x)` on the same graph with a different color and label.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
plt.plot(x, np.sin(x), label='sin(x)', color='steelblue')
plt.plot(x, np.cos(x), label='cos(x)', color='darkorange')
plt.xlabel('x')
plt.ylabel('value')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Scatter Plot

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = 300 * np.random.rand(50)
colors = np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, alpha=0.6, cmap='viridis')
plt.title("Scatter Plot Example")
plt.xlabel("x")
plt.ylabel("y")
plt.colorbar(label="color scale")
plt.show()
```

✅ **Your Turn**: Generate two sets of random points and plot them in different colors on the same scatter plot.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(4543)
x1, y1 = rng.normal(0, 1, 50), rng.normal(0, 1, 50)
x2, y2 = rng.normal(3, 1, 50), rng.normal(2, 1, 50)
plt.scatter(x1, y1, color='teal', label='Group A')
plt.scatter(x2, y2, color='coral', label='Group B')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Bar Plot

```python
categories = ["A", "B", "C", "D"]
values = [5, 7, 3, 8]

plt.bar(categories, values)
plt.title("Bar Plot Example")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

✅ **Your Turn**: Create your own bar chart with at least 4 categories (e.g., favorite foods, courses, or hobbies).

```python
import matplotlib.pyplot as plt

categories = ['Python', 'GIS', 'SQL', 'Stats']
hours = [7, 5, 4, 3]
plt.bar(categories, hours, color=['#2a9d8f', '#457b9d', '#e9c46a', '#e76f51'])
plt.ylabel('Study hours')
plt.title('Weekly Study Focus')
plt.show()
```

## 5. Customizing Plots

```python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin(x)", linestyle="--", color="blue")
plt.plot(x, y2, label="cos(x)", linestyle=":", color="red")
plt.title("Customized Line Plot")
plt.xlabel("x (radians)")
plt.ylabel("value")
plt.legend()
plt.grid(True)
plt.show()
```

✅ **Your Turn**: Customize your own plot with labels, legend, and grid lines.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 8)
y = x ** 2
plt.plot(x, y, marker='o', linewidth=2, label='x squared')
plt.xlabel('x value')
plt.ylabel('y value')
plt.title('Customized Square Plot')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

### Summary
- `%matplotlib inline` makes plots display inside the notebook.
- Line, scatter, and bar plots are the most common basics.
- Titles, labels, legends, and grids improve readability.
- Visuals tell stories about data more effectively than raw numbers.
